# Exercise 1
### Write code to print the first 10 natural numbers to the console using a while loop.

n=1

while n<=10 do

    print(n)

    n = n + 1

end while

```python
def print_n_to_m(n, m):
    while n <= m:
        print(n, end=" ")
        n += 1

n=1
m=10

print_n_to_m(n, m)  

print("\n")

print_n_to_m(1, 5)

# Das definieren von Funktionen hat den Vorteil, dass man den Code nicht immer wieder neu schreiben muss. 
# Man kann die Funktion einfach aufrufen und die gewünschten Parameter übergeben.
```

## Write code to print all prime numbers between any given pair of numbers n and m to the console.
Use as example input n = 25 and m = 50.
NOTE:A prime number is a natural number greater than 1 that can only be divided by itself and
1 without remainders. Examples are: 2, 3, 5, 7, 11.


number ← 25
m ← 50


while number ≤ m do

    divisor ← 2
    prime ← true

    while divisor < number do

        if number mod divisor = 0 then
            prime ← false
        end if

        divisor ← divisor + 1

    end while

    if prime = true then
        print number
    end if

    number ← number + 1

end while

```python
number=0
m=50

if number < 2:
    number = 2

while number <= m:
    divisor = 2
    prime = True

    while divisor * divisor <= number:
        if number % divisor == 0:
            prime = False
            break
        divisor += 1

    if prime:
        print(number, end=" ")

    number += 1

```

```python
start=0 # inclusive
stop=10 # exclusive
step=1 # step size

for number in range(start, stop, step):

    if number < 2:
        continue

    divisor = 2
    prime = True

    while divisor * divisor <= number:
        if number % divisor == 0:
            prime = False
            break
        divisor += 1

    if prime:
        print(number, end=" ")
```
